```python
#create an api client 
from anthropic import Anthropic


from helper import (
    add_user_message,
    add_assistant_message,
    chat
)

client = Anthropic()
model = "claude-sonnet-5-5"
```

```python
#make an initial list of messages 
messages = []
#system prompts
system =  """
you are a patient math tutor, do not directly awnser a student awnser, instead walk them 
through the solution step by step 
"""

add_user_message(messages, "how to solve 2+2")
awnser = chat(messages, system = system)

awnser
```
